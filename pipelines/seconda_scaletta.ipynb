# Analisi delle Traiettorie Evolutive Stratificate per Clock Rank

L'obiettivo di questa analisi è ricostruire un albero evolutivo "stratificato" che caratterizzi uno specifico tipo di tumore, partendo dai dati di diversi pazienti. L'analisi si basa sul concetto di **Clock Rank**, dove ogni mutazione è assegnata a una finestra temporale (rank) specifica, permettendo di mappare la progressione tumorale dall'inizio (Rank 1) alle fasi successive.

## 1. Selezione della Coorte e Scoperta dei Candidati

- **Filtraggio per Tipo di Tumore**: Isolamento di tutti i campioni appartenenti allo stesso `ttype` (es. ESAD, BRCA).
- **Ricerca Esplorativa dei Driver**: Invece di usare una lista predefinita, si identificano i geni o i segmenti genomici più ricorrenti nella coorte. L'obiettivo è scoprire quali alterazioni caratterizzano il tumore analizzato.
- **Mappatura del Timing**: Estrazione della colonna `clock_rank` per ogni alterazione candidata per stabilire la sua posizione temporale relativa (Rank 1 = precoce, Rank 2+ = tardivo).

## 2. Analisi della Distribuzione Temporale e Recorrenza

- **Mappatura Gene-Rank**: Calcolare la distribuzione di frequenza di ogni gene nei diversi `clock_rank`. Un gene che appare costantemente al Rank 1 in molti pazienti è un forte candidato come driver d'inizio.
- **Analisi di Co-occorrenza**: Verificare quali geni tendono a comparire insieme (stesso rank) o in sequenza fissa tra pazienti diversi.

## 3. Identificazione dei Driver Caratterizzanti

- **Significatività per Layer**: Definire i geni "driver" del tumore in base al loro impatto evolutivo:
    - **Iniziatori (Rank 1)**: Geni ricorrenti che appaiono nelle fasi più precoci.
    - **Progressori (Rank 2+)**: Geni che guidano l'evoluzione successiva e l'espansione clonale.
- **Filtro di Rilevanza**: Mantenere solo i geni che mostrano una ricorrenza statistica significativa nella coorte.

## 4. Costruzione dell'Albero Stratificato

- **Architettura a Livelli**: Costruzione di un grafo dove i livelli (Y-axis) sono determinati dal `clock_rank`.
- **Nodi**: Rappresentano i geni scoperti come rilevanti per quel tipo di tumore.
- **Archi**: Connessioni tra geni di rank diversi che appaiono sequenzialmente negli stessi pazienti, evidenziando il "percorso tipico" di evoluzione del tumore.

## 5. Visualizzazione e Validazione

- **Layout Gerarchico**: Visualizzazione chiara dei layer temporali.
- **Interpretazione**: Identificazione dei geni "hub" (fondamentali per la progressione) e delle varianti evolutive meno comuni.
- **Conclusione**: L'albero finale rappresenta la firma evolutiva temporale che caratterizza il tipo di tumore studiato.
